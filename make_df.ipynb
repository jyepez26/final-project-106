```python
import pandas as pd
```

```python
start_cd = 'all_data/wearable_data/Data/'
```

```python
students = ['S'+str(i) for i in range(1,11)]
```

```python
tests = ['Final', 'Midterm 1', 'Midterm 2']
```

```python
ACC_df_prep = pd.DataFrame()
for s in students:
    for t in tests:
        df_paused = pd.read_csv(start_cd+s+'/'+t+'/ACC.csv')
        df_paused = df_paused.drop(0).reset_index().drop(columns=['index'])
        cols = df_paused.columns
        df_paused = df_paused.rename(columns={cols[0]:'x',cols[1]:'y',cols[-1]:'z'})
        df_paused['student'] = s
        df_paused['test'] = t
        ACC_df_prep = pd.concat([ACC_df_prep,df_paused])

ACC_df = ACC_df_prep
ACC_df.head()
```

```python
BVP_df_prep = pd.DataFrame()
for s in students:
    for t in tests:
        df_paused = pd.read_csv(start_cd+s+'/'+t+'/BVP.csv')
        df_paused = df_paused.drop(0).reset_index().drop(columns=['index'])
        cols = df_paused.columns
        df_paused = df_paused.rename(columns={cols[0]:'data'})
        df_paused['student'] = s
        df_paused['test'] = t
        BVP_df_prep = pd.concat([BVP_df_prep,df_paused])

BVP_df = BVP_df_prep
BVP_df
```

```python
BVP_df_m1 = BVP_df[BVP_df['test'] == 'Midterm 1'].drop(columns=['test'])
BVP_df_m2 = BVP_df[BVP_df['test'] == 'Midterm 2']
BVP_df_fi = BVP_df[BVP_df['test'] == 'Final']
```

```python
BVP_df_m1['row'] = BVP_df_m1.groupby('student').cumcount()
```

```python
BVP_df_m1_pivot = BVP_df_m1.pivot(index='row', columns='student', values='data')
BVP_df_m1_pivot.reset_index(drop=True, inplace=True)
BVP_df_m1_pivot
```

```python
BVP_df_m2['row'] = BVP_df_m2.groupby('student').cumcount()
```

```python
BVP_df_m2_pivot = BVP_df_m2.pivot(index='row', columns='student', values='data')
BVP_df_m2_pivot.reset_index(drop=True, inplace=True)
BVP_df_m2_pivot
```

```python
BVP_df_fi['row'] = BVP_df_fi.groupby('student').cumcount()
```

```python
BVP_df_fi_pivot = BVP_df_fi.pivot(index='row', columns='student', values='data')
BVP_df_fi_pivot.reset_index(drop=True, inplace=True)
BVP_df_fi_pivot
```
